```python
import pandas as pd
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import pypickle
```

```python
ds=pd.read_csv("./datasets/placement_data.csv")
```

```python
ds
```

```python
ds.shape
```

```python
ds["Placement"].value_counts()
```

```python
xfeature=ds[["CGPA","AptitudeScore","CommunicationSkills","Internship","Projects","Other Skills"]]
ytarget=ds[["Placement"]]
```

```python
model=LogisticRegression()
```

```python
model.fit(xfeature,ytarget)
ypp=model.predict(xfeature)
accuracy_score(ytarget,ypp)*100
```

```python
pypickle.save("placement_model.pkl",model)
```
